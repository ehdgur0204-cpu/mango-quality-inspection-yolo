```python
# subprocess: 학습 스크립트를 별도 프로세스로 실행하기 위한 모듈
import os
import subprocess
import sys
from pathlib import Path

# 현재 작업 폴더에서 프로젝트 루트를 찾는다
repo_root = Path.cwd()
if not (repo_root / "models").is_dir():
    repo_root = repo_root.parent

# 동일한 조건으로 학습할 YOLO 모델 목록
models = ["yolov8n", "yolov8s", "yolov8m"]
train_script = repo_root / "models" / "base_model" / "train_baseline.py"

# Python과 Ultralytics 로그가 즉시 출력되도록 버퍼링을 비활성화한다
environment = {**os.environ, "PYTHONUNBUFFERED": "1"}

# 모델을 하나씩 순서대로 학습한다
for model_name in models:
    print(f"\n===== Training {model_name} =====", flush=True)

    # 학습 프로세스를 시작하고 표준 출력과 오류 출력을 함께 받는다
    process = subprocess.Popen(
        [sys.executable, "-u", str(train_script), "--model", model_name],
        cwd=repo_root,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        bufsize=0,
        env=environment,
    )

    # YOLO의 Epoch 및 batch 진행률을 한 글자씩 실시간으로 표시한다
    while True:
        output = process.stdout.read(1)
        if not output:
            break
        sys.stdout.write(output.decode("utf-8", errors="replace"))
        sys.stdout.flush()

    # 학습 프로세스가 정상 종료되었는지 확인한다
    return_code = process.wait()
    if return_code != 0:
        raise RuntimeError(f"{model_name} training failed with exit code {return_code}")
```
